### Calcultion with Variables

```python
original_price = 7000
price_after_discount = 6760

# What is the discount percentage?
discount_percentage = ((original_price - price_after_discount) / original_price) * 100
discount_percentage
```

```python
discount = (discount_percentage * original_price) / 100
discount
```
