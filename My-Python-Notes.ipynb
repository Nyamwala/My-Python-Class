### Dictionaries
- It is pythons most powerful built in data type
- it allows you to store data as key value pairs


## Comparison to lists
- in a list you access elements using an index, which must be an integer
- in a dictionary the indices can be (almost) any type. we call these indices keys
- Each key maps to a specific value

Terminology
- Key; The label or index (a word or user ID)
- Value; The data associated with the key (a definition or a score)
- Item; The combination of a key and its value


### Creating and populating dictionaries

> you can create dictionaries using the dict() constructor or curly braces {}

### English to spanish mapping

```python
# 1. Creating an empty dictionary
eng2sp = dict()
print(eng2sp)
# output: {} <--- curly braces represent a dictionary

# 2. Addding items (key value pairs)
# syntax dictionary_name[key] = value
eng2sp["one"] = "uno"
eng2sp["two"] = "dos"

print(eng2sp)
# output: {'one': 'uno', 'two': 'dos'}
```

### Initialization syntax 
- you can also create a dictionary with data already inside it
- The output shown above is exactly the same as the input format

```python
eng2sp = {'one': 'uno', 'two': 'dos', 'three': 'tres'}
```

### Essential properties and operations 
- Dictionaries preserve insertion order. If you add "one" then "two" they will stay in that order

- length (len)
- The len function counts the number of key value pairs 

```python
print(len(eng2sp))
# output: 3
```

- The in operator (Membership)


- The in operator checks if something exists as a key not as a value


```python
# Checking keys
print("one" in eng2sp) # True
print("uno" in eng2sp) # False (because "uno" is a value, not a key)
```

> To check if something exists as a value you must use the .value() method

```python
vals = list(eng2sp.values()) # converts values into a list ['uno', 'dos', 'tres']
print('uno' in vals) # True
```

### Method A: The explicit logic (using if/else)
- we iterate through a string for each character
1. check if we have seen it before (is it in the dictionary)
2. if No: Create a new entry with count 1
3. if yes: Update the existing entry by adding 1

```python
word = "brontosaurus"
d = dict()
for c in word:
    if c not in d:
        d[c] = 1 # first time seeing this letter
else:
    d[c] = d[c] + 1 # seen it before, increment the count by 1
    print(d)
    # output: {'b': 1, 'r': 2, 'o': 2, 'n': 1, 't': 1, 's': 2, 'a': 1, 'u': 2}
```

### Loops

- we have the following type of loops in python
    - for loops
    - while loops
    

**for loops**

```python
student_grades = [10, 13, 65, 59, 21, 45, 90, 33, 0, 88 ]
```

```python
print(student_grades)
```

```python
# printing the grade of each student in the list
for grade in student_grades:
    print(grade)
```

**conditional statements in for loops**

```python
# print out only grades that are greater than or equal to 50

for grade in student_grades:
    # check if the grade is greater than or equal to 50
    if grade >= 50:
        print(grade)
```

```python
# check how many students have scored 50 and above

graduating_students = 0
students_not_graduating = 0
for grade in student_grades:
    # check if the grade is greater than or equal to 50
    if grade >= 50:
        graduating_students += 1
    else:
        students_not_graduating +=1

print(f"number of students who are graduating {graduating_students}")
print(f"number of student not graduating:{students_not_graduating}")
print(f"Total number of students: {graduating_students + students_not_graduating}")
```

### while loops


- 'break' and 'continue' statement

- while loops are very useful for continuous tasks :
    - Accelerate 'while' a button is being pressed
    - monitor 'while' below or above threshhold

```python
# items in stock
stock = 10

# Number of purchases 
purchases = 0

while purchases < stock:
    # when a customer buys something increase purchases by 1
    purchases += 1
    # print for me remaining stock from each purchase
    print(f"remaining stock: {stock - purchases}")
```
