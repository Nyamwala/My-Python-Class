```python
class CustomerGikosh:
    """A class that will be  a blueprint for all the customers in Gikosh
    It will contain the following states and behaviour
    - States: name, rewards_number, email, phone_number
    - Behaviour: make_purchase, return_purchase, order_online, cancel_order
    """

    def __init__(self, name, email, rewards_number, phone_number ):

        self.name = name 
        self.rewards_number = rewards_number
        self.email = email
        self.phone_number = phone_number

    def make_purchase(self): 
          print(f"Thank You {self.name} for your purchase. You have earned 200 points under reward number {self.rewards_number}")
    def return_purchase(self):
          print("return of purchse for order 092 is approved. You will receive a refund in 3-5 business days.")
    def order_online(self):
          print(f"Thank you for your online order {self.name}")
    def cancel_order(self):
          print(f"Your order for {self.name} has been cancelled")
```

## Inheritance

```python
class BankAccount:
    """ This is a class used to instantiate Bank customers
    states: balance"""
    
    def __init__(self, balance):
        self.balance = balance

    def withdraw(seff, amount):
        self.balance -= amount

# An empty class inheriting from the BankAccount class

class SavingsAccount(BankAccount):
    # constructor for savings account with additional arguments \
    def __init__(self, balance, interest_rate):
        # Call the parent constructor using ClassName.__init__()
        # Self is a savings account but also a BankAccount
        BankAccount.__init__(self, balance)

        # Add more functionality
        self.interest_rate = interest_rate
    def compute_interest(self, n_period = 1):
        interest = self.balance * ((1 + self.interest_rate) ** n_period -1)
        return interest + self.balance
```

```python
acct = SavingsAccount(4000, 0.3)
acct.interest_rate
```

```python
acct.compute_interest(4)
```

```python
class BankAccount:
    """A class used to instantiate bank customers."""

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        self.balance -= amount


class SavingsAccount(BankAccount):
    """A savings account that earns simple interest."""

    def __init__(self, balance, interest_rate):
        super().__init__(balance)
        self.interest_rate = interest_rate

    def compute_interest(self, n_period=1):
        # Simple interest final balance: P * (1 + r * t)
        interest = self.balance * self.interest_rate * n_period
        return interest
```

```python
acct = SavingsAccount(4000, 0.3)
acct.interest_rate
```

```python
acct.compute_interest(4)
```
