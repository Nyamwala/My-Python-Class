```python
import numpy as np

# np_students has a shape, (5000, 2)
# column 0 = height (m), column 1 = weight (kg)

# 1. Weight above 40 kg -> std_40
std_40 = np_students[np.where(np_students[:, 1] > 40)[0]]

# 2. Height shorter than or equal to the mean height
mean_height = np_students[:, 0].mean()
shorter_students = np_students[np.where(np_students[:, 0] <= mean_height)[0]]

# 3. Sort std_40 by height, lowest to highest
std_40 = std_40[np.argsort(std_40[:, 0])]

# 4. Delete the first 20 rows of std_40
std_40 = np.delete(std_40, np.arange(20), axis=0)

# 5. Add a random names column to np_students
first_names = ["Roberto", "Brian", "Yasmin", "Erick", "Esther",
               "Dennis", "Grace", "Hamza", "Irene", "James"]
names = np.random.choice(first_names, size=np_students.shape[0])   # shape (5000,)
names = names.reshape(-1, 1)                                       # shape (5000, 1)
np_students = np.concatenate((np_students, names), axis=1)         # shape (5000, 3)

print(np_students[:5])
print(std_40.shape, shorter_students.shape, np_students.shape)
```
