**Creating arrays from scratch**


- There are three major ways to create arrays: 
    - "np.zeros": This creates an array of zeros
    - 'np.rando.random' : This creates an array with random numbers
    - 'np.arange' : Creates an array with values in a specified range.

```python
# creating an array with 'np.zeros'
np.zeros((5,3))
```

```python
# creating an array with 'np.random.random'
np.random.random((10, 4))
```

```python
# Creating an array with np.arange

np.arange(50, 60)
```

```python
plt.scatter(np.arange(0, 7), # We have 14 elements
            np.arange(-3, 4))  # we have 14 elements
plt.show()
```
