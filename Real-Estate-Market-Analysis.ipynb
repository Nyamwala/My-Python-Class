### Real Estate Market Analysis

```python
import numpy as np

# 1. Load the dataset as a structured array using your file name
file_name = 'Real-estate-dataset.csv'

data = np.genfromtxt(
    file_name,
    delimiter=',',  # Tells NumPy columns are separated by commas
    names=True,  # Uses the first row of your CSV as the column/field headers
    dtype=None,  # Automatically detects data types (numbers, text, etc.)
    encoding=None,  # Properly handles text encoding
)

# 2. Inspect the shape (shows how many rows/properties you have)
print('Dataset Shape:', data.shape)

# 3. Inspect the fields (column names)
print('\nFields (Columns):', data.dtype.names)

# 4. Inspect data types for each field
print('\nData Types:')
print(data.dtype)

# 5. Display the first five records to verify the data looks correct
print('\nFirst 5 Records:')
print(data[:5])
```

**Analyze Property Prices**

```python
# Extract price column (replace 'Price' with the exact column name from your dataset if different)
prices = data['Price']

mean_price = np.mean(prices)
median_price = np.median(prices)
min_price = np.min(prices)
max_price = np.max(prices)
percentiles = np.percentile(prices, [25, 50, 75])

print(f'Mean Price: ${mean_price:,.2f}')
print(f'Median Price: ${median_price:,.2f}')
print(f'Minimum Price: ${min_price:,.2f}')
print(f'Maximum Price: ${max_price:,.2f}')
print(
    f'25th, 50th, 75th Percentiles: ${percentiles[0]:,.2f},'
    f' ${percentiles[1]:,.2f}, ${percentiles[2]:,.2f}'
)
```

**Analyze Property Size**

```python
# Extract size column (replace 'SquareFeet' with your dataset's column name)
sizes = data['Square_Feet']

mean_size = np.mean(sizes)
median_size = np.median(sizes)
min_size = np.min(sizes)
max_size = np.max(sizes)

print(f'Mean Property Size: {mean_size:,.2f} sq ft')
print(f'Median Property Size: {median_size:,.2f} sq ft')
print(f'Minimum Property Size: {min_size:,.2f} sq ft')
print(f'Maximum Property Size: {max_size:,.2f} sq ft')
```

**Calculate Price per Square Foot**

```python
# Element-wise calculation
price_per_sqft = data['Price'] / data['Square_Feet']

pps_mean = np.mean(price_per_sqft)
pps_median = np.median(price_per_sqft)
pps_min = np.min(price_per_sqft)
pps_max = np.max(price_per_sqft)

print(f'Mean Price per Sq Ft: ${pps_mean:.2f}')
print(f'Median Price per Sq Ft: ${pps_median:.2f}')
print(f'Min Price per Sq Ft: ${pps_min:.2f}')
print(f'Max Price per Sq Ft: ${pps_max:.2f}')
```

**Identify Expensive Properties**

```python
# Find the 75th percentile threshold
p75 = np.percentile(data['Price'], 75)

# Create a boolean mask for top 25% properties
expensive_mask = data['Price'] >= p75
expensive_props = data[expensive_mask]

num_expensive = expensive_props.size
avg_expensive_price = np.mean(expensive_props['Price'])

print(f'Number of properties in top 25%: {num_expensive}')
print(f'Average price of top 25%: ${avg_expensive_price:,.2f}')
print(f'Overall average price: ${mean_price:,.2f}')
```

**Apply Multiple Conditions**

```python
avg_size = np.mean(data['Square_Feet'])

# Combine conditions: Larger than average AND top 25% price AND Location score >= 8
# (Make sure column names match your dataset: e.g., 'Square_Feet', 'Price', 'Location_Score')
multi_condition_mask = (
    (data['Square_Feet'] > avg_size)
    & (data['Price'] >= p75)
    & (data['Location_Score'] >= 8)
)

filtered_properties = data[multi_condition_mask]

print(f'Number of properties meeting all conditions: {filtered_properties.size}')
print('\nSample of matching properties:')
print(filtered_properties[:5])
```

**Compare Prices to the Average**

```python
# Broadcasting: subtract scalar mean_price from every element in prices array
price_diff = data['Price'] - mean_price

above_avg_count = np.sum(price_diff > 0)
below_avg_count = np.sum(price_diff < 0)

print(f'Properties above average price: {above_avg_count}')
print(f'Properties below average price: {below_avg_count}')
print(
    'Explanation: NumPy broadcasting automatically expands the scalar mean price'
    ' across the entire array, allowing element-wise subtraction in a single'
    ' vector operation without explicit loops.'
)
```

**Find the 10 Most Expensive Properties**

```python
# Sort indices based on price
sorted_indices = np.argsort(data['Price'])

# Get indices of the 10 most expensive properties (last 10 elements)
top_10_indices = sorted_indices[-10:]

# Use fancy indexing to retrieve records
top_10_properties = data[top_10_indices]

# Display details (adjust field names like 'ID', 'Square_Feet', 'Bedrooms', 'Location_Score', 'Price' as needed)
for row in top_10_properties:
  print(
      f"ID: {row['ID']} | Size: {row['Square_Feet']} sq ft | Bedrooms:"
      f" {row['Num_Bedrooms']} | Location_Score: {row['Location_Score']} | Price:"
      f" ${row['Price']:,.2f}"
  )
```

**Final Analysis**

```python
# Calculate metrics for the top 10 properties
top_10_size = np.mean(top_10_properties['Square_Feet'])
top_10_beds = np.mean(top_10_properties['Num_Bedrooms'])
top_10_loc = np.mean(top_10_properties['Location_Score'])

overall_beds = np.mean(data['Num_Bedrooms'])
overall_loc = np.mean(data['Location_Score'])

print(
    f'Top 10 Average Size: {top_10_size:.1f} sq ft (Overall:'
    f' {mean_size:.1f} sq ft)'
)
print(
    f'Top 10 Average Num_Bedrooms: {top_10_beds:.1f} (Overall:'
    f' {overall_beds:.1f})'
)
print(
    f'Top 10 Average Location_Score: {top_10_loc:.1f} (Overall:'
    f' {overall_loc:.1f})'
)
```
